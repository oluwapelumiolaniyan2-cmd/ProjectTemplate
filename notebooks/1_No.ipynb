```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# Preprocess education group using one-hot encoding
preprocessor_spai = ColumnTransformer(
    transformers=[
        ("education", OneHotEncoder(drop="first", handle_unknown="ignore"),
         ["education_group"])
    ]
)

# Create Linear Regression pipeline
spai_model = Pipeline(steps=[
    ("preprocessor", preprocessor_spai),
    ("regressor", LinearRegression())
])

# Train the model
spai_model.fit(X_train, y_train_spai)

# Make predictions
spai_predictions = spai_model.predict(X_test)
```

```python
print("SPAI Linear Regression Results")

print("R²:", round(r2_score(y_test_spai, spai_predictions), 3))

print("RMSE:", round(
    np.sqrt(mean_squared_error(y_test_spai, spai_predictions)), 3
))

print("MAE:", round(
    mean_absolute_error(y_test_spai, spai_predictions), 3
))
```

```python
# Preprocess education group
preprocessor_nmp = ColumnTransformer(
    transformers=[
        ("education", OneHotEncoder(drop="first", handle_unknown="ignore"),
         ["education_group"])
    ]
)

# Create Linear Regression pipeline
nmp_model = Pipeline(steps=[
    ("preprocessor", preprocessor_nmp),
    ("regressor", LinearRegression())
])

# Train the model
nmp_model.fit(X_train, y_train_nmp)

# Make predictions
nmp_predictions = nmp_model.predict(X_test)
```

```python
print("NMP Linear Regression Results")

print("R²:", round(r2_score(y_test_nmp, nmp_predictions), 3))

print("RMSE:", round(
    np.sqrt(mean_squared_error(y_test_nmp, nmp_predictions)), 3
))

print("MAE:", round(
    mean_absolute_error(y_test_nmp, nmp_predictions), 3
))
```

```python
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

RANDOM_STATE = 42

# Individual models
def create_logistic():
    return Pipeline([
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
        ("model", LogisticRegression(max_iter=1000))
    ])

def create_random_forest():
    return Pipeline([
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
        ("model", RandomForestClassifier(
            n_estimators=200,
            max_depth=3,
            min_samples_leaf=10,
            random_state=RANDOM_STATE
        ))
    ])

def create_knn():
    return Pipeline([
        ("encoder", OneHotEncoder(handle_unknown="ignore")),
        ("model", KNeighborsClassifier(n_neighbors=5))
    ])

# Soft Voting Ensemble
def create_ensemble():
    return VotingClassifier(
        estimators=[
            ("lr", create_logistic()),
            ("rf", create_random_forest()),
            ("knn", create_knn())
        ],
        voting="soft"
    )
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
```

```python
def create_preprocessor(numerical_columns, categorical_columns):

    numerical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            drop="first"
        ))
    ])

    return ColumnTransformer([
        ("numerical", numerical_pipeline, numerical_columns),
        ("categorical", categorical_pipeline, categorical_columns)
    ])
```

```python
# Model 1: NMP only
model1 = Pipeline([
    ("preprocessor", create_preprocessor(
        ["NMP_score"], []
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Model 2: NMP + SAPS
model2 = Pipeline([
    ("preprocessor", create_preprocessor(
        ["NMP_score", "SAPS_score"], []
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Model 3: NMP + SAPS + demographics
model3 = Pipeline([
    ("preprocessor", create_preprocessor(
        ["NMP_score", "SAPS_score"],
        ["gender", "grade"]
    )),
    ("classifier", LogisticRegression(max_iter=1000))
])
```

```python
model1.fit(X1_train, y_train)
model2.fit(X2_train, y_train)
model3.fit(X3_train, y_train)
```

```python
predictions = pd.DataFrame({
    "Actual category": y_test,
    "Model 1": model1.predict(X1_test),
    "Model 2": model2.predict(X2_test),
    "Model 3": model3.predict(X3_test)
})

print(predictions.head(10))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

models = {
    "Model 1: NMP only": (model1, X1_test),
    "Model 2: NMP + SAPS": (model2, X2_test),
    "Model 3: Full model": (model3, X3_test)
}

results = []

for name, (model, X_test) in models.items():

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results)

print(results_df.round(3))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = [
    "accuracy",
    "precision",
    "recall",
    "f1",
    "roc_auc"
]

models = {
    "Model 1: NMP only": model1,
    "Model 2: NMP + SAPS": model2,
    "Model 3: Full model": model3
}

cv_results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X1_train if name.startswith("Model 1") else
        X2_train if name.startswith("Model 2") else X3_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    cv_results.append({
        "Model": name,
        "Accuracy": scores["test_accuracy"].mean(),
        "Precision": scores["test_precision"].mean(),
        "Recall": scores["test_recall"].mean(),
        "F1-score": scores["test_f1"].mean(),
        "ROC-AUC": scores["test_roc_auc"].mean()
    })

cv_results_df = pd.DataFrame(cv_results)

print(cv_results_df.round(3))
```
