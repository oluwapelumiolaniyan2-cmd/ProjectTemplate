```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# Preprocess education group using one-hot encoding
preprocessor_spai = ColumnTransformer(
    transformers=[
        ("education", OneHotEncoder(drop="first", handle_unknown="ignore"),
         ["education_group"])
    ]
)

# Create Linear Regression pipeline
spai_model = Pipeline(steps=[
    ("preprocessor", preprocessor_spai),
    ("regressor", LinearRegression())
])

# Train the model
spai_model.fit(X_train, y_train_spai)

# Make predictions
spai_predictions = spai_model.predict(X_test)
```

```python
print("SPAI Linear Regression Results")

print("R²:", round(r2_score(y_test_spai, spai_predictions), 3))

print("RMSE:", round(
    np.sqrt(mean_squared_error(y_test_spai, spai_predictions)), 3
))

print("MAE:", round(
    mean_absolute_error(y_test_spai, spai_predictions), 3
))
```

```python
# Preprocess education group
preprocessor_nmp = ColumnTransformer(
    transformers=[
        ("education", OneHotEncoder(drop="first", handle_unknown="ignore"),
         ["education_group"])
    ]
)

# Create Linear Regression pipeline
nmp_model = Pipeline(steps=[
    ("preprocessor", preprocessor_nmp),
    ("regressor", LinearRegression())
])

# Train the model
nmp_model.fit(X_train, y_train_nmp)

# Make predictions
nmp_predictions = nmp_model.predict(X_test)
```

```python
print("NMP Linear Regression Results")

print("R²:", round(r2_score(y_test_nmp, nmp_predictions), 3))

print("RMSE:", round(
    np.sqrt(mean_squared_error(y_test_nmp, nmp_predictions)), 3
))

print("MAE:", round(
    mean_absolute_error(y_test_nmp, nmp_predictions), 3
))
```

```python
results = pd.DataFrame({
    "Model": ["Smartphone Addiction", "Nomophobia"],
    "R²": [
        r2_score(y_test_spai, spai_predictions),
        r2_score(y_test_nmp, nmp_predictions)
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test_spai, spai_predictions)),
        np.sqrt(mean_squared_error(y_test_nmp, nmp_predictions))
    ],
    "MAE": [
        mean_absolute_error(y_test_spai, spai_predictions),
        mean_absolute_error(y_test_nmp, nmp_predictions)
    ]
})

print(results.round(3))
```
